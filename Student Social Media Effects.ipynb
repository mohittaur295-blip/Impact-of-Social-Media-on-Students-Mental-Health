```python
df['Gender']=df['Gender'].fillna('unknown')
df['Academic_Level']=df['Academic_Level'].fillna('unknown')
df['Country']=df['Country'].fillna('Country')
df['Most_Used_Platform']=df['Most_Used_Platform'].fillna(df['Most_Used_Platform'].mode()[0])
df['Purpose_Of_Use']=df['Purpose_Of_Use'].fillna(df['Purpose_Of_Use'].mode()[0])
df['Stress_Level']=df['Stress_Level'].fillna(df['Stress_Level'].mode()[0])
```

```python
df['Age']=df['Age'].fillna(df['Age'].median())
df['Avg_Daily_Usage_Hours']=df['Avg_Daily_Usage_Hours'].fillna(df['Avg_Daily_Usage_Hours'].median())
df['Daily_Unlocks']=df['Daily_Unlocks'].fillna(df['Daily_Unlocks'].median())
df['Sleep_Hours_Per_Night']=df['Sleep_Hours_Per_Night'].fillna(df['Sleep_Hours_Per_Night'].median())
df['Study_Hours']=df['Study_Hours'].fillna(df['Study_Hours'].median())
df['Physical_Activity_Hours']=df['Physical_Activity_Hours'].fillna(df['Physical_Activity_Hours'].median())
df['Mental_Health_Score']=df['Mental_Health_Score'].fillna(df['Mental_Health_Score'].median())
```

```python
def detect_outliers_iqr(df, column):
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1
    lb = q1 - 1.5 * iqr
    ub = q3 + 1.5 * iqr
    outliers = df[(df[column] < lb) | (df[column] > ub)]
    return outliers

numeric_cols = ['Age','Avg_Daily_Usage_Hours','Daily_Unlocks','Study_Hours','Physical_Activity_Hours',
                'Sleep_Hours_Per_Night','Mental_Health_Score']

for col in numeric_cols:
    out = detect_outliers_iqr(df, col)
    print(f"{col}: {len(out)} outliers")

outlier_indices = set()
for col in numeric_cols:
    outlier_indices.update(detect_outliers_iqr(df, col).index)

df_clean = df.drop(index=outlier_indices)
print("Original:", df.shape, "-> Cleaned:", df_clean.shape)
df=df_clean
```
